```python
basic_salary = int(input("enter the salary: "))
allowance = int(input("enter your allowance: "))

gross_salary = basic_salary + allowance

if (gross_salary > 80000):
  tax = gross_salary * 0.15
  print("tax above 80000: ", tax)
elif (gross_salary > 50000 and gross_salary <= 80000):
  tax = gross_salary * 0.1
  print("tax between 50000 and 80000: ", tax)
elif (gross_salary  >30000 and gross_salary <= 50000):
  tax = gross_salary * 0.5
  print("tax between 30000 and 50000: ", tax)
else:
  tax = 0
  print("no tax added: ", tax)

net_salary = gross_salary - tax
print("net salary: ", net_salary)
```
